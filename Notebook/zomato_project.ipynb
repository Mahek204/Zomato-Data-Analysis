```python
def rate_clean(str_data):
  data_array= str_data.split('/')
  rate_str= data_array[0] #we need only first part of data after spliting rate into 2 parts
  return float(rate_str)
data["rate"]=data["rate"].apply(rate_clean)
```

This function cleans restaurant ratings like "4.1/5" in dataset and converts them into a numeric float (4.1) to perform numerical analysis (e.g., finding averages).

```python
pivot_table = data.pivot_table(index='listed_in(type)', columns='online_order', aggfunc='size', fill_value=0)
print(pivot_table)
sns.heatmap(pivot_table, annot=True,cmap="YlGnBu", fmt='d')
plt.title("Heatmap")
plt.xlabel("Online Order")
plt.ylabel("Listed In (Type)")
plt.show()
```

1.

*Online ordering is more common among Cafes and small outlets, while Dining restaurants mostly rely on in-person customers.

*This suggests that traditional dinning still dominates, but digital ordering is gaining traction, especially in Cafes and niche outlets.

```python
#correlation heatmap
sns.heatmap(data.corr(numeric_only=True), annot=True, cmap="coolwarm")
plt.title("Correlation Between Numeric Features")
plt.show()
```

This creates a square grid where:

Each row and column represents a numeric feature (like rate, votes, approx_cost(for two people)).

Each cell shows how strongly two variables are correlated.


rate ↔ votes:	+0.6 to +0.8

    Restaurants with more votes generally have higher ratings. Popularity and rating go hand-in-hand.

rate ↔ approx_cost(for two people):	~0.1 or -0.1

    Very weak relationship which means that price doesn’t strongly affect ratings.
votes ↔ approx_cost(for two people): 	+0.3

    Slight positive correlation which means that expensive restaurants might attract more votes.
